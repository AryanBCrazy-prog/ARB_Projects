```python
# OMG a Rocket Launch 
#thank you for reading this idk made this from scratch
#Used time module though confused on how to slow the text speed.
fuel_Level = float(input("Enter your fuel level (in %tage , from 1 to 100): "))
cargo_Weight = float(input("Enter your cargo weight(in kgs) "))
weather_is_clear = input("do you want the weather clear (y/n): ").strip().lower() in ("yes", "y", "true")
engine_Temperature = float(input("Enter engine temperature(in degree celsius): "))
countdown = int(input("Start Countdown in: "))
launch_approved = False

if fuel_Level>= 80 and cargo_Weight<= 5000 and weather_is_clear == True :
    print ("All systems green. Clear for launch.")
    launch_approved = True
elif fuel_Level>= 80 and cargo_Weight>= 5000 and weather_is_clear == True :
    print ("Cargo too heavy. Launch aborted.")
elif fuel_Level>= 80 and cargo_Weight<= 5000 and weather_is_clear == False :
    print ("Launch delayed due to weather.")
else :
    print ("Pre-flight check failed. Maintenance required. Fuel level too low")

import time
time.sleep(2)
if launch_approved == True:
    print ("Proceeding for lanching")
    time.sleep(0.5)
    while countdown > 0:
        time.sleep(1)
        countdown = countdown - 1
        print (countdown)

    print ("LIFTOFF !!")
    time.sleep(3)
    for layer in range (4):
        engine_Temperature = engine_Temperature + 15
        time.sleep(1)
        print ("Layer", layer + 1, ": Engine temp is", engine_Temperature)
        
        if engine_Temperature >= 90:
            print("Warning , Engine Overheating!")
            time.sleep(1)
            print("This is the end....... Hold your breath and count to ten , Feel the Earth move, and then KABOOM")
            break
    time.sleep(2)
    print("Flight simulation ended.")
else:
    time.sleep(1)
    print ("Simulation aborted. The rocket is staying on the ground, You stupid.")
time.sleep(1)
print("Flight simulation ended.")
time.sleep(1)
print("First ever Python long project by Aryan , appreciate me")

```
